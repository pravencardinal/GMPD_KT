```python
from google.cloud.storage import Client as StorageClient
from ingestion.connectors.sharepoint import SharepointConnector, SharepointType
from ingestion.models.config import GCSStorageConfig, MicrosoftGraphConfig
from ingestion.utils.logging import setup_logger
from ingestion.storage.discovery_engine import get_or_create_engine
import logging
import nest_asyncio
nest_asyncio.apply()
```

```python
logger = logging.getLogger(__name__)
setup_logger(logger)
logger.info("Setting up...")
```

```python
storage_client = StorageClient()
location = "global"
domain = "GMPD-Masterdata"
data_store_id = "domain-gmpd-masterdata" 

# Sharepoint
title1 = "GlobalProductDeployment"
title2 = "ProductLifecycleManagement"
title3 = "IMPDTraining"
title4 = "CMDKnowledgeBase"
title5 = "ProductDatabaseMedical"
drive_folder = None
```

```python
connector1 = SharepointConnector(
            project=storage_client.project,
            domain=domain,
            data_store_id=data_store_id,
            site_title=title1,
            gcs_storage_config=GCSStorageConfig(project=storage_client.project),
            graph_config=MicrosoftGraphConfig(project=storage_client.project),
        )

connector1.index(logger=logger, drive_folders=drive_folder)
connector1.mirror_drive_files(logger=logger)
connector1.mirror_site_pages(logger=logger)
```

```python
connector2 = SharepointConnector(
            project=storage_client.project,
            domain=domain,
            data_store_id=data_store_id,
            site_title=title2,
            gcs_storage_config=GCSStorageConfig(project=storage_client.project),
            graph_config=MicrosoftGraphConfig(project=storage_client.project),
        )

connector2.index(logger=logger, drive_folders=drive_folder)
connector2.mirror_drive_files(logger=logger)
connector2.mirror_site_pages(logger=logger)
```

```python
connector3 = SharepointConnector(
            project=storage_client.project,
            domain=domain,
            data_store_id=data_store_id,
            site_title=title3,
            gcs_storage_config=GCSStorageConfig(project=storage_client.project),
            graph_config=MicrosoftGraphConfig(project=storage_client.project),
        )

connector3.index(logger=logger, drive_folders=drive_folder)
connector3.mirror_drive_files(logger=logger)
connector3.mirror_site_pages(logger=logger)
```
